```python
import pandas as pd
import joblib
```

```python
scaler = joblib.load("/kaggle/input/notebooks/shreyaashambhavii/predict-customer-churn-notebook-2/scaler.pkl")
model  = joblib.load("/kaggle/input/notebooks/shreyaashambhavii/predict-customer-churn-notebook-2/log_reg.pkl")
```

```python
test = pd.read_csv("/kaggle/input/notebooks/shreyaashambhavii/predict-customer-churn-notebook-1/test.csv")
```

```python
X_test      = scaler.transform(test)
churn_proba = model.predict_proba(X_test)[:, 1]
```

```python
test_raw = pd.read_csv("/kaggle/input/competitions/playground-series-s6e3/test.csv")
```

```python
submission = pd.DataFrame({"id": test_raw["id"], "Churn": churn_proba})
```

```python
submission.to_csv("submission.csv", index = False)

print(submission.head(10))
```
