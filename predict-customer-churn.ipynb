```python
train = train_dataset.copy()
test = test_dataset.copy()
```

```python
train = train.drop(columns = ["id"])
test = test.drop(columns = ["id"])
```

```python
train = train.rename(columns = {'gender': 'Gender'})
test = test.rename(columns = {'gender': 'Gender'})
```

```python
train['Gender'] = train['Gender'].map({'Male': 0, 'Female': 1})
test['Gender'] = test['Gender'].map({'Male': 0, 'Female': 1})
```

```python
train = train.rename(columns = {'tenure': 'Tenure'})
test = test.rename(columns = {'tenure': 'Tenure'})
```

```python
bins = [0, 12, 24, 36, 48, 60, 72]
labels = [1, 2, 3, 4, 5, 6]

train['Tenure'] = pd.cut(train['Tenure'], bins = bins, labels = labels)
test['Tenure'] = pd.cut(test['Tenure'], bins = bins, labels = labels)
```
